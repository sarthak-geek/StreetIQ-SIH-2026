```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
import matplotlib.patches as patches
from google.colab import drive
```

```python
drive.mount('/content/drive')
```

```python
path = '/content/drive/MyDrive/SIH_dataset/dataset/train'
```

```python
labels = pd.read_csv(os.path.join(path,'_annotations.csv'))
```

```python
print(labels.shape)
labels.head()
```

```python
images = []
for filename in os.listdir(path):
  if filename == '_annotations.csv':
    continue
  img = Image.open(os.path.join(path, filename))
  images.append([filename, img])
```

```python
i= 0
img_name = images[i][0]
img = images[i][1]
fig, ax = plt.subplots()
potholes = labels[labels['filename'] == img_name]
plt.imshow(img)

print(img.size)
potholes
for _, row in potholes.iterrows():
  xmin = row['xmin']
  xmax = row['xmax']
  ymin = row['ymin']
  ymax = row['ymax']
  w = row['width']
  h = row['height']
  rect = patches.Rectangle((xmin,ymin), (xmax-xmin), (ymax-ymin),edgecolor='red',linewidth=2,alpha=0.4)
  ax.add_patch(rect)
plt.show()
```
