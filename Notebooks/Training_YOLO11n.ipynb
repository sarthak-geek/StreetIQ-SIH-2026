```python
# Import
from ultralytics import YOLO
import os
from google.colab import drive
```

```python
drive.mount('/content/drive')
```

```python
# Dataset
dataset_path = "/content/drive/MyDrive/SIH_dataset/yolo_dataset"
yaml_path = f"{dataset_path}/data.yaml"
# Load pretrained YOLO11 Nano
model = YOLO("yolo11n.pt")
```

```python
# Train
results = model.train(
    data=yaml_path,
    epochs=50,
    imgsz=640,
    batch=16,
    device=0,
    project="/content/drive/MyDrive/SIH_dataset/yolo_training",
    name="pothole_yolo11n"
)
```
